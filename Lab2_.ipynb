```python
# Method 1: Delete Rows with Missing Values

# dropna() removes all rows containing missing values.

df_drop = df.dropna()

print("\n========== Shape After Removing Missing Rows ==========")
print(df_drop.shape)
```

```python
# Method 2: Replace Missing Age with Mean
# Make a copy of the dataset first.

df_mean = df.copy()

# Fill missing Age values using Mean

df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

print("\n========== Missing Values After Mean ==========")
print(df_mean["Age"].isnull().sum())
```

```python
# Method 3: Replace Missing Age with Median
# Create another copy of the dataset.

df_median = df.copy()

# Fill missing Age values using Median

df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

print("\n========== Missing Values After Median ==========")
print(df_median["Age"].isnull().sum())
```

```python
# Method 4: Replace Missing Embarked with Mode
# Create another copy of dataset.

df_mode = df.copy()

# mode()[0] returns the most frequently occurring value.

df_mode["Cabin"] = df_mode["Cabin"].fillna(df_mode["Cabin"].mode()[0])

print("\n========== Missing Values After Mode ==========")
print(df_mode["Cabin"].isnull().sum())
```

```python
# Check missing values after preprocessing.

print("\n========== Missing Values (Mean Method) ==========")
print(df_mean.isnull().sum())

print("\n========== Missing Values (Median Method) ==========")
print(df_median.isnull().sum())

print("\n========== Missing Values (Mode Method) ==========")
print(df_mode.isnull().sum())
```
